```python
import json
import csv
import requests

# Load JSON from GitHub
url = "https://raw.githubusercontent.com/KilowattApp/open-ev-data/master/data/ev-data.json"
resp = requests.get(url)
data = resp.json()["data"]  # Only the "data" list

# Prepare rows for CSV
rows = []
for car in data:
    variant = car.get("variant") or "unknown"
    release_year = car.get("release_year") or "unknown"

    row = {
        "id": car.get("id"),
        "brand": car.get("brand"),
        "brand_id": car.get("brand_id"),
        "model": car.get("model"),
        "vehicle_type": car.get("vehicle_type"),
        "variant": variant,
        "release_year": release_year,
        "usable_battery_size": car.get("usable_battery_size"),
        "average_consumption": car.get("energy_consumption", {}).get("average_consumption"),
        "ac_charger_ports": ",".join(car.get("ac_charger", {}).get("ports", [])),
        "ac_charger_usable_phases": car.get("ac_charger", {}).get("usable_phases"),
        "ac_charger_max_power": car.get("ac_charger", {}).get("max_power"),
        "dc_charger_ports": ",".join(car.get("dc_charger", {}).get("ports", [])),
        "dc_charger_max_power": car.get("dc_charger", {}).get("max_power"),
        "is_default_charging_curve": car.get("dc_charger", {}).get("is_default_charging_curve"),
        "charging_voltage": car.get("charging_voltage"),
        "charging_curve": json.dumps(car.get("dc_charger", {}).get("charging_curve", []))
    }
    rows.append(row)

# Write CSV
fieldnames = [
    "id","brand","brand_id","model","vehicle_type","variant","release_year",
    "usable_battery_size","average_consumption",
    "ac_charger_ports","ac_charger_usable_phases","ac_charger_max_power",
    "dc_charger_ports","dc_charger_max_power","is_default_charging_curve",
    "charging_voltage","charging_curve"
]

with open("open_ev_data_1.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(rows)

print(f"Saved {len(rows)} EV entries to open_ev_data.csv")
```
